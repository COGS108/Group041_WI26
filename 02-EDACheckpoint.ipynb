# Research Question

Is there any influence of biological factors and physical attributes (including height, age, and handedness) on game success (measured by players ranking and tennis match wins) the most among the 2024 professional ATP and WTA tennis players?

## Background and Prior Work

As multiple people in our group play tennis, we decided to focus our project on which biological factors influence the win rate of professional tennis players. Looking at factors such as height, weight, dominant hand preference, wingspan, among other factors, we hope to determine which attributes seem to most heavily influence winning rates for top players. While we recognize that factors such as coaching, years of experience, court type, and more also influence a players performance, we aim to focus on which physical characteristics of the players influence performance.

Winning in tennis is attributed to an interaction between training regimen, technical skill, and psychological, and biological factors. Among these, biological factors in the field of tennis is an interesting field of research as it is the least malleable trait. Genetic dispositions such as length of legs could positively or adversely affect span and balance. The scale of how impactful biology can be to the field of tennis is evident by studying handedness. On average, the frequency of the population expressing left handedness is 10%. Among the top 100 professional players, that number skyrockets to 20%<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1).

While they found that shorter players tend to have a higher return rate on clay courts, taller players have a higher serve percentage on all types of courts<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2). While this study mentions shorter players returning strength and taller players serving rates, how does height impact the remainder of the player’s performance for the remainder of the shots in a point? How large is the impact of serve and return rates on overall match winning rate? This analysis determines which types of court surfaces and winning rates affect serve and return rates, and how height plays a role on different types of courts. Our study aims to look at how height affects winning rates regardless of court type and serve and return percentages. Additionally, our project aims to determine if other biological factors impact player performance as well. This study was helpful in seeing how a similar study to our question was constructed and what metrics were used in their analysis, and we are looking into a question that is looking at more biological factors and simply winning rate. While we recognize from this study that height impacts serve and return rates, we will explore how height impacts overall performance.

### ATP Tennis Dataset
1. The Jeff Sackmann ATP Tennis dataset is a dataset containing records of professional men’s tennis matches played on the ATP tour. Each row in the dataset represents one completed match and includes information about players, match outcomes, and tournament details. It also includes a csv that details all given players personal statistics including but not limited to: height, age, handedness, win rate, and rounds played.

    The key dependent variable we will be using in the dataset is match outcome, usually represented by winner and loser identifiers. A score is a series of points players have, with the first player reaching 6 being the winner. For example, the scores can be something like 6-2 or 6-4 depending on how each player does. Additionally, some matches on the dataset have scores such as 7-5, indicating a tie-breaker occurred in the set. A tie-breaker occurs when both players reach 5-5. We will be using scores to determine the match outcome, using binary code to identify wins and losses. For example, a 1 would signify a win while a 0 would signify a loss. These will act as our dependent variables in the study. 

    The independent variables are players’ physical attributes and demographic characteristics, which we will be measuring to determine if these factors influence performance outcomes in professional tennis. These variables include height, handedness, and age. Each metric captures a distinct dimension of physical advantage or developmental stage that could affect win rate.

    Height is recorded in centimeters (cm) in the ATP dataset. Professional male tennis players typically range from approximately 165 cm to over 210 cm in height. Height is especially relevant in tennis due to its impact on serving. Taller players can generate sharper serve angles and higher contact points, which greatly influence aces occurring in the match. An ace is when a player serves in a way that prevents the returner from even coming into contact with the ball. However, extreme height may reduce agility or endurance. With all of this being said, we want to analyze players to see if height acts as a benefit or drawback when playing tennis. If there is a benefit, what height makes being tall a “drawback” while playing tennis.

    Handedness is an independent variable indicating whether a player is right-handed or left-handed. A majority of tennis players are right-handed, with left-handed players representing a minority (typically around 10-15% of the professional population). Handedness is a strategic key while playing tennis since most players are trained to play against right-handed players. Left-handed players often create unusual spin patterns and serve angles, which become especially challenging for players trained to expect the opposite. We will be measuring handedness through a binary variable (0 = right-handed, 1 = left-handed).

    Age is calculated in years. Players generally enter the professional tennis league in their late teens, peak physically and competitively in their early-to-mid 20s, and experience performance decline in their 30s due to reduced recovery capacity and increased injury risk. With that being said, older players do have several advantages over younger players. For instance, older players have a plethora of experience and strategic intelligence, which is correlated with age. Therefore, age may exhibit a normal distribution with win rate rather than a strictly positive or negative one. These independent variables capture structural physical and biological characteristics that may influence match outcomes. These 3 independent variables will help signify the role physicality plays in tennis and if it affects match outcomes. 

2. The most glaring problem of this datset is thjat it contains exclusively professional ATP and WTA tour-level matches in 2024. This means it only represents elite players who qualified for major tournaments. This introduces a drawback known as survirvorship bias. This concept describes how lower-ranked players are not sampled and included in the dataset. Additionally, tennis can be a costly and expensive sport. This means it is a globally uneven sport sp the sample overrepresents players from wealtheir countries that could support stronger traininf for their ahtletes that compete. Because the dataset only coevers one season, short term injuries and schedule differences in the 2024 season can distort data predicting long term performance predictions. Lastly, contextrual variables such as injuries, coaching, and fatigue both mental and physical are missing considerations. Therefore, findings that we attain from this datset should state it only applies to eite professonals in tennis for 2024 and has ot be careful when generalized to broader tennis populations. 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# A. Loading the Dataset:
mens_df = pd.read_csv("data/00-raw/atp_matches_2024.csv")

# B. Tidying the dataset:
# drop unneccessary columns
mens_df = mens_df[['winner_name', 'winner_hand', 'winner_ht', 'winner_age', 'score', 
                'tourney_name', 'minutes', 'loser_name', 'loser_hand', 'loser_ht', 
                'loser_age', 'winner_rank', 'winner_rank_points', 'loser_rank', 
                'loser_rank_points']]

# C. dataset size:
print('Size of the ATP mens datset: ', mens_df.shape, '\n')

# D find missing data:
# drop matches that are forfited:
mens_df = mens_df[mens_df['score'] != 'W/O']
# drop matches that were retired/RET (ended early for injury):
mens_df = mens_df[~mens_df['score'].str.contains('RET')]

# where is data missing at this point:
print('Missing data per column:')
missing = mens_df.isnull().sum()
print(missing, '\n')

missing_info = mens_df.isnull().any(axis=1)
missing_info = mens_df[missing_info.eq(True)]
print('Tournaments where there are the most missing values:')
print(missing_info['tourney_name'].value_counts())
```

From the above tables, we can see that most null values are for minutes and from the Roland Garros, Paris Olympics, and Davis Cup tournament. We can conclude that there is some systematic issue with these specific tournaments and their data collection.

```python
# F. clean data, drop the rows where player info is not recorded
# Since this information is critical to our analysis, we need data where all this information is intact
mens_df = mens_df.dropna()

# E find outliers:
# describe variables where it is possible to identify outliers
print('Summary statistics of variables to identify outliers:')
print(mens_df.describe())
print(mens_df.describe(include='object'))
```

From this, we can see that there are only 2 unique values for dominant hand preference. The unique values for the dominant hand preference are:

```python
print('Unique hand values: ', mens_df['winner_hand'].unique())
print(mens_df.groupby('winner_hand')['loser_hand'].value_counts())
```

There are no ambidextrous (U) men in this data, so we don't have to drop them.

We also see that there ares some strange minimum values in some of the data. We will visualize it with boxplots: 

```python
fig, axes = plt.subplots(1, 2, figsize=(8, 3))
axes[0].boxplot([mens_df['winner_age'], mens_df['loser_age']], labels=['Winner', 'Loser'])
axes[0].set_ylabel('Age')

axes[1].boxplot([mens_df['winner_ht'], mens_df['loser_ht']],labels=['Winner', 'Loser'])
axes[1].set_ylabel('Height (cm)')

plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(8, 3))
axes[0].boxplot(mens_df['minutes'], labels=['Game'])
axes[0].set_ylabel('minutes')

axes[1].boxplot([mens_df['winner_rank_points'], mens_df['loser_rank_points']],labels=['Winner', 'Loser'])
axes[1].set_ylabel('Rank Points')

plt.tight_layout()
plt.show()
```

From these plots we can identify a clear outlier in the loser heights at 71 cm. We should remove this row if this player has an incorrectly recorded height. Lets identify them:

```python
print(mens_df[mens_df['loser_ht'] < 100][['loser_ht', 'loser_name']])
```

```python
mens_df = mens_df[mens_df['loser_ht'] > 100]
plt.figure(figsize=(8,3))
plt.boxplot([mens_df['winner_ht'], mens_df['loser_ht']],labels=['Winner', 'Loser'])
plt.ylabel('Height (cm)')
print('Data without outlier:')
plt.show()
```

The rest of this data seems like there is no clear outliers that are impossible to come accross in a tennis match. Although we see alot of outliers in the upper regions of the age and rank points chart, this doesn't cause alarm because of the structure of tennis ranking and the fact that many successful tennis players continue to play later in life.

Since there are significantly less left handed players, we need to look at the porportion of games won instead of the count. 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

palette = {
    'male': 'cornflowerblue',
    'female': 'coral'
}

df_mens = pd.read_csv('data/02-processed/atp_matches_2024.csv')
df_womens = pd.read_csv('data/02-processed/wta_matches_2024.csv')
df_mens['gender'] = 'male'
df_womens['gender'] = 'female'
df_all = pd.concat([df_mens, df_womens], ignore_index=True)
```

```python
wins = df_all.groupby(['gender', 'winner_hand']).size()
total_played = df_all.groupby(['gender', 'winner_hand']).size() + df_all.groupby(['gender', 'loser_hand']).size()
prop_df = (wins / total_played).reset_index(name="proportion_won")
sns.barplot(prop_df, x='winner_hand', y="proportion_won", hue='gender', palette=palette)
plt.title('Win Rate for Left and Right Handed Players')
```

As we can see here, theres not much difference at all between the win rates of left and right handed players for Men. For women, we see that left handed players have a lower win rate.

Now we want to see if its possible that a left handed player can have an advantage playing against a right handed player or vice versa. We will investigate this by looking at the left and right handed player's win proportion when playing against the opposite handed player. 

```python
lr_matches = df_all[
    ((df_all['winner_hand'] == 'L') & (df_all['loser_hand'] == 'R')) |
    ((df_all['winner_hand'] == 'R') & (df_all['loser_hand'] == 'L'))
]
lr_wins = lr_matches.groupby(['gender', 'winner_hand']).size()
lr_total = lr_matches.groupby('gender').size()

lr_prop = lr_wins.div(lr_total, level='gender').reset_index(name='proportion_won')
sns.barplot(data=lr_prop, x='winner_hand', y='proportion_won', hue='gender', palette=palette)
plt.title('Win Rate for Left Vs. Right Handed Matches')
```

From this we can see that right handed women have a even higher win porportion against left handed women. For men, there is no significant difference again. 

#### Age

Next, we will explore the relationship (again, if any) between age and ranking points. We will first look at the general distribution of our wrangled and cleaned data to get a better understanding of the data we are working with. For this part, we are keeping ATP and WTA data separate, to see if the influence of age on rank points differs by gender as well.

```python
# Set up 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

atp = pd.read_csv("data/02-processed/atp_matches_2024.csv")
wta = pd.read_csv("data/02-processed/wta_matches_2024.csv")

### Check Shape of Age Distribution
fig, axes = plt.subplots(1, 4)
fig.subplots_adjust(hspace=0.5, wspace=0.5, top=0.9, bottom=0.1, left=0.01, right=0.99)

atp['winner_age'].hist(bins=20, ax=axes[0], figsize=(12, 4))
atp['loser_age'].hist(bins=20, ax=axes[1])
axes[0].set_title('ATP Winner Ages'), axes[1].set_title('ATP Loser Ages')

wta['winner_age'].hist(bins=20, ax=axes[2])
wta['loser_age'].hist(bins=20, ax=axes[3])
axes[2].set_title('WTA Winner Ages'), axes[3].set_title('WTA Loser Ages')

#### Check Shape of Rank Points Distribution
fig, axes = plt.subplots(1, 4)
fig.subplots_adjust(hspace=0.5, wspace=0.5, top=0.9, bottom=0.1, left=0.01, right=0.99)

atp['winner_rank_points'].hist(bins=20, ax=axes[0], figsize=(12, 4))
atp['loser_rank_points'].hist(bins=20, ax=axes[1])
axes[0].set_title('ATP Winner Rank Points'), axes[1].set_title('ATP Loser Rank Points')

wta['winner_rank_points'].hist(bins=20, ax=axes[2])
wta['loser_rank_points'].hist(bins=20, ax=axes[3])
axes[2].set_title('WTA Winner Rank Points'), axes[3].set_title('WTA Loser Rank Points')
```

We can see that all of the age distributions are approximately normally distributed, with a slight skew to the right. The distribtuion of Rank Points, however, are all skewed to the right significantly, indicating that most players fall in the same range of rank points with some top players having significantly more points. \
Now that we have checked the shape of the distribution for age and rank points, we will explore whether there is a relationship between the two. To do this, we will be looking at the difference between the winner's and loser's ages and the difference between their rank points. 

```python
### Calculate differences and store columns of interest in new df
atp['age_diff'] = atp['winner_age'] - atp['loser_age']
atp['rank_points_diff'] = atp['winner_rank_points'] - atp['loser_rank_points']

wta['age_diff'] = wta['winner_age'] - wta['loser_age']
wta['rank_points_diff'] = wta['winner_rank_points'] - wta['loser_rank_points']

# Plot to inspect relationship between diffrences
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

atp.plot.scatter(x='age_diff', y='rank_points_diff', ax=ax1, title='ATP Age Difference vs Rank Points Difference')
wta.plot.scatter(x='age_diff', y='rank_points_diff', ax=ax2, title='WTA Age Difference vs Rank Points Difference')
```

As we can see, there is no correlation between age difference and rank points difference. This means that age likely does not make a significant impact on rank points. While we cannot conclude that age does not impact a players win rate at all based on this alone, we can clearly see that the difference in two player's age is not correlated with their rank points difference. Lastly, we will look at a pairplot of the players ages and rank points to look for any other potential correlation, and as represented in the plot below, there is no correlation.


```python
wta_pairplot = wta[['winner_age', 'winner_rank_points', 'loser_age', 'loser_rank_points']]
sns.pairplot(wta_pairplot)
```
