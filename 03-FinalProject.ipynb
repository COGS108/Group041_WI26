#### Handedness

Let us first investigate the effect (if any) on Handedness on winning the match. We want to see if theres any difference between how many games won by right handed vs. left handed people. Since there are significantly less left handed players, we need to look at the porportion of games won instead of the count. 

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import pearsonr

palette = {
    'male': 'cornflowerblue',
    'female': 'coral'
}

df_mens = pd.read_csv('data/02-processed/atp_matches_2024.csv')
df_womens = pd.read_csv('data/02-processed/wta_matches_2024.csv')
df_mens['gender'] = 'male'
df_womens['gender'] = 'female'
df_all = pd.concat([df_mens, df_womens], ignore_index=True)
```

```python
wins = df_all.groupby(['gender', 'winner_hand']).size()
total_played = df_all.groupby(['gender', 'winner_hand']).size() + df_all.groupby(['gender', 'loser_hand']).size()
prop_df = (wins / total_played).reset_index(name="proportion_won")
sns.barplot(prop_df, x='winner_hand', y="proportion_won", hue='gender', palette=palette)
plt.title('Win Rate for Left and Right Handed Players')
```

As we can see here, theres not much difference at all between the win rates of left and right handed players for Men. For women, we see that left handed players have a lower win rate.

Now we want to see if its possible that a left handed player can have an advantage playing against a right handed player or vice versa. We will investigate this by looking at the left and right handed player's win proportion when playing against the opposite handed player. 

```python
lr_matches = df_all[
    ((df_all['winner_hand'] == 'L') & (df_all['loser_hand'] == 'R')) |
    ((df_all['winner_hand'] == 'R') & (df_all['loser_hand'] == 'L'))
]
lr_wins = lr_matches.groupby(['gender', 'winner_hand']).size()
lr_total = lr_matches.groupby('gender').size()

lr_prop = lr_wins.div(lr_total, level='gender').reset_index(name='proportion_won')
sns.barplot(data=lr_prop, x='winner_hand', y='proportion_won', hue='gender', palette=palette)
plt.title('Win Rate for Left Vs. Right Handed Matches')
```

From this we can see that right handed women have a even higher win porportion against left handed women. For men, there is no significant difference again. 

```python

df_all = pd.read_csv("data/00-raw/atp_matches_2024.csv")
winner_df = df_all[['winner_hand', 'winner_rank_points']].rename(
    columns={
        'winner_hand': 'hand',
        'winner_rank_points': 'rank_points'
    }
)
loser_df = df_all[['loser_hand', 'loser_rank_points']].rename(
    columns={
        'loser_hand': 'hand',
        'loser_rank_points': 'rank_points'
    }
)
combined_df = pd.concat([winner_df, loser_df])
combined_df = combined_df.dropna()
combined_df = combined_df[combined_df['hand'].isin(['L','R'])]
avg_df = combined_df.groupby('hand')['rank_points'].mean().reset_index()
plt.figure(figsize=(8,6))
sns.barplot(
    data=avg_df,
    x='hand',
    y='rank_points'
)

plt.title("Average ATP Ranking Points by Handedness (2024)")
plt.xlabel("Handedness (L = Left, R = Right)")
plt.ylabel("Average Ranking Points")

plt.show()
```

This further proves that ranking points and handedness have a low correlation. We can see that a right handed player has higher rank points, on average, however we still cannot draw any strong conclusions about their correlations since there are also significantly more right handed players than left handed.

```python
atp = pd.read_csv("data/02-processed/atp_matches_2024.csv")
wta = pd.read_csv("data/02-processed/wta_matches_2024.csv")

### Check Shape of Age Distribution
fig, axes = plt.subplots(1, 4)
fig.subplots_adjust(hspace=0.5, wspace=0.5, top=0.9, bottom=0.1, left=0.01, right=0.99)

atp['winner_age'].hist(bins=20, ax=axes[0], figsize=(12, 4))
atp['loser_age'].hist(bins=20, ax=axes[1])
axes[0].set_title('ATP Winner Ages'), axes[1].set_title('ATP Loser Ages')

wta['winner_age'].hist(bins=20, ax=axes[2])
wta['loser_age'].hist(bins=20, ax=axes[3])
axes[2].set_title('WTA Winner Ages'), axes[3].set_title('WTA Loser Ages')

#### Check Shape of Rank Points Distribution
fig, axes = plt.subplots(1, 4)
fig.subplots_adjust(hspace=0.5, wspace=0.5, top=0.9, bottom=0.1, left=0.01, right=0.99)

atp['winner_rank_points'].hist(bins=20, ax=axes[0], figsize=(12, 4))
atp['loser_rank_points'].hist(bins=20, ax=axes[1])
axes[0].set_title('ATP Winner Rank Points'), axes[1].set_title('ATP Loser Rank Points')

wta['winner_rank_points'].hist(bins=20, ax=axes[2])
wta['loser_rank_points'].hist(bins=20, ax=axes[3])
axes[2].set_title('WTA Winner Rank Points'), axes[3].set_title('WTA Loser Rank Points')
```

We can see that all of the age distributions are approximately normally distributed, with a slight skew to the right. The distribtuion of Rank Points, however, are all skewed to the right significantly, indicating that most players fall in the same range of rank points with some top players having significantly more points. \
Now that we have checked the shape of the distribution for age and rank points, we will explore whether there is a relationship between the two. To do this, we will be looking at the difference between the winner's and loser's ages and the difference between their rank points. 

```python
### Calculate differences and store columns of interest in new df
atp['age_diff'] = atp['winner_age'] - atp['loser_age']
atp['rank_points_diff'] = atp['winner_rank_points'] - atp['loser_rank_points']

wta['age_diff'] = wta['winner_age'] - wta['loser_age']
wta['rank_points_diff'] = wta['winner_rank_points'] - wta['loser_rank_points']

# Plot to inspect relationship between diffrences
fig, (ax1, ax2) = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

sns.regplot(data=atp, x='age_diff', y='rank_points_diff', ax=ax1, line_kws={"color": "darkblue"})
title = ax1.set_title('ATP Age Difference vs Rank Points Difference')
sns.regplot(data=wta, x='age_diff', y='rank_points_diff', ax=ax2, line_kws={"color": "darkblue"})
title = ax2.set_title('WTA Age Difference vs Rank Points Difference')

#Pearson's R:
atp_coefficient, atp_p_value = pearsonr(atp['age_diff'], atp['rank_points_diff'])
print('ATP Pearson\'s coefficient: ', atp_coefficient)
print('ATP Pearson\'s R: ', atp_p_value)

wta_coefficient, wta_p_value = pearsonr(wta['age_diff'], wta['rank_points_diff'])
print('WTA Pearson\'s coefficient: ', wta_coefficient)
print('WTA Pearson\'s R: ', wta_p_value)
```

As we can see, there is no correlation between age difference and rank points difference. The Pearson's R is very small for both ATP and WTA, indicating that age likely does not make a significant impact on rank points. While we cannot conclude that age does not impact a players win rate at all based on this alone, we can clearly see that the difference in two player's age is not correlated with their rank points difference. Lastly, we will look at a pairplot of the players ages and rank points to look for any other potential correlation, and as represented in the plot below, there is no correlation.


```python
wta_pairplot = wta[['winner_age', 'winner_rank_points', 'loser_age', 'loser_rank_points']]
sns.pairplot(wta_pairplot)
```

#### Height
Next, we will explore the relationship between height and ranking points. We will look and see if height has any effect on rank points. We will be also keeping this section separated by gender, as well as height.

```python
atp =pd.read_csv("data/02-processed/atp_matches_2024.csv")
wta = pd.read_csv("data/02-processed/wta_matches_2024.csv")

atp_winners = atp[['winner_ht', 'winner_rank_points']].rename(columns={'winner_ht': 'Height','winner_rank_points': 'Points'})
atp_losers =atp[['loser_ht', 'loser_rank_points']].rename(columns={'loser_ht':'Height','loser_rank_points': 'Points'})
men = pd.concat([atp_winners,atp_losers])
men['Gender'] = 'Men'

wta_winners = wta[['winner_ht', 'winner_rank_points']].rename(columns={'winner_ht': 'Height', 'winner_rank_points': 'Points'})
wta_losers = wta[['loser_ht', 'loser_rank_points']].rename(columns={'loser_ht': 'Height', 'loser_rank_points': 'Points'})
women = pd.concat([wta_winners, wta_losers])
women['Gender'] = 'Women'

df = pd.concat([men, women]).dropna()
```

```python
plt.figure(figsize=(12, 12))
sns.lmplot(data=df, x='Height', y='Points', hue='Gender', height=5, aspect=1.8)

plt.title('Height vs. Rank Points (Men and Women)', fontsize=15)
plt.xlabel('Height (cm)', fontsize=12)
plt.ylabel('Rank Points', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

#Pearson's R:
atp_coefficent, atp_p_value = pearsonr(men['Height'], men['Points'])
print('Mens Pearson\'s coefficient: ', atp_coefficient)
print('Mens Pearson\'s R: ', atp_p_value)

wta_coefficient, wta_p_value = pearsonr(women['Height'], women['Points'])
print('Womens Pearson\'s coefficient: ', wta_coefficient)
print('Womens Pearson\'s R: ', wta_p_value)
```

We can see that there is not a strong correlation between Height and Rank Points regardless of gender. We can see the middle portion of the distribution for each respective gender favoring a more "average" height for the gender. We can see outliers for extremely tall and short players for both genders each, but they are rarely scoring a high number of rank points. Being an "average" seems to be more favorable for both genders.
Additionally, the Pearson's coefficient and R suggest no significant relationship, likely due to the unusual shape of the data.

```python
men_losers = atp[['loser_ht','loser_rank_points']].rename(columns={'loser_ht': 'Height', 'loser_rank_points': 'Points'})
men_losers['Gender'] = 'Men'
women_losers = wta[['loser_ht','loser_rank_points']].rename(columns={'loser_ht': 'Height', 'loser_rank_points': 'Points'})
women_losers['Gender'] ='Women'
df_losers =pd.concat([men_losers, women_losers]).dropna()

plt.figure(figsize=(12, 7))

sns.lmplot(data=df_losers,x='Height', y='Points', hue='Gender', palette='Set2', height=5, aspect=1.8)
plt.title('Losers Height vs. Rank Points (ATP & WTA)',fontsize=15)
plt.xlabel('Height (cm)',fontsize=12)
plt.ylabel('Rank Points', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

#Pearson's R:
atp_coefficent, atp_p_value = pearsonr(men_losers['Height'], men_losers['Points'])
print('Mens Pearson\'s coefficient: ', atp_coefficient)
print('Mens Pearson\'s R: ', atp_p_value)

wta_coefficient, wta_p_value = pearsonr(women_losers['Height'], women_losers['Points'])
print('Womens Pearson\'s coefficient: ', wta_coefficient)
print('Womens Pearson\'s R: ', wta_p_value)
```

Here, we can see that the distribution looks similar to that of the Winners, except overall height of the distribution is a little less. The overall relationship between Height and Rank Points scored is consistent between winners and losers, showing us that the "peak" players are usually of "average" (middle of each gender distribution) height. 
Again, Pearson's R and coefficient suggest no significant relationship.
